```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y = make_regression(n_samples = 4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

```python
# Lets apply the OLS
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(X,y)
```

```python

reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red')
```

```python
# Lets apply Gradient Descent assuming slope is constant m = 78.35
# and let's assume the starting value for intercept b = 0

y_pred = ((78.35*X ) +100).reshape(4)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='#00a65a',label='b = 0')
plt.legend()
plt.show()
```

```python
m = 78.35
b = 100

loss_slope = -2 * np.sum(y-m*X.ravel() - b)
loss_slope
```

```python
# lets take the lerning rate = 0.1\

lr = 0.1


step_size = loss_slope * lr
step_size
```

```python
# Calculating the new intercept

b = b - step_size
b
```

```python
m = 78.35
b_initial = 100
y_pred_initial = ((m*X) + b_initial).reshape(4)

# The current value of 'b' in the kernel is the updated one
y_pred_updated = ((m*X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred_initial,color='#00a65a',label='b = {}'.format(b_initial))
plt.plot(X,y_pred_updated,color='#A3E4D7',label='b = {}'.format(b))
plt.legend()
plt.show()
```

```python
# Iteration 2
loss_slope = -2 * np.sum(y-m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b-step_size
b
```

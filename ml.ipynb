```python
import pandas as pd
from flask import Flask, request, jsonify
from flask_cors import CORS
from sklearn.ensemble import RandomForestRegressor

app = Flask(__name__)
CORS(app)

# Load your Kaggle dataset
df = pd.read_csv('Smart_Rainwater_Harvesting_Dataset.csv')

# Train ML Model (Predicting Water Collected)
X = df[['Roof_Area(sqm)', 'Rainfall_mm', 'Storage_Tank_Capacity_Liters']]
y = df['Water_Collected_Liters']
model = RandomForestRegressor(n_estimators=100).fit(X, y)

@app.route('/predict', methods=['POST'])
def predict():
    data = request.json
    roof = float(data.get('roof'))
    rain = float(data.get('rain'))
    cap = float(data.get('capacity'))
    
    # ML Prediction for Water
    water_pred = model.predict([[roof, rain, cap]])[0]
    
    # Integrated Solar Logic (Less rain = More sun)
    solar_gen = max(2, 25 - (rain * 0.4))
    
    return jsonify({
        "water": round(water_pred, 1),
        "solar": round(solar_gen, 1),
        "percent": round((water_pred / cap) * 100, 1),
        "overflow": "High" if water_pred >= cap else "Low"
    })

if __name__ == '__main__':
    app.run(port=5000, debug=True)
```
